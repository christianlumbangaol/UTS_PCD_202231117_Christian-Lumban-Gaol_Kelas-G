```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img_bgr = cv2.imread('GAMBAR_BACKLIGHT.jpeg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
```

```python
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
```

### Operasi Pixel

```python
def adjust_logic(image, brightness=0, contrast=1.0):
    res = image.astype(np.float32)
    res = res * contrast + brightness
    res = np.clip(res, 0, 255).astype(np.uint8)
    return res
```

```python
bright_val = 80
contrast_val = 1.8

img_brightened = adjust_logic(img_gray, brightness=bright_val, contrast=1.0)
img_contrasted = adjust_logic(img_gray, brightness=0, contrast=contrast_val)
img_final      = adjust_logic(img_gray, brightness=bright_val, contrast=contrast_val)
```

```python
titles = [
    'GAMBAR ASLI', 
    'GAMBAR GRAY', 
    'GAMBAR GRAY YANG DIPERCERAH', 
    'GAMBAR GRAY YANG DIPERKONTRAS', 
    'GAMBAR GRAY DIPERCERAH DAN DIPERKONTRAS'
]
images = [img_rgb, img_gray, img_brightened, img_contrasted, img_final]

plt.figure(figsize=(20, 12))

for i in range(5):
    plt.subplot(2, 3, i+1)
    if i == 0:
        plt.imshow(images[i])
    else:
        plt.imshow(images[i], cmap='gray')
    plt.title(titles[i], fontweight="bold")
    plt.axis('off')

plt.tight_layout()
plt.show()
```
